```python
import pandas as pd
import numpy as np
import matplotlib.pylab as plt
from scipy.interpolate import interp1d
```

```python
df = pd.read_csv('craterscale.dat', sep='\s+')
```

```python
df
```

```python
plt.plot(df['#Dimp(m)'], df['Dcrat(m)'])
```

```python
Dikm = df['#Dimp(m)'] / 1000
Dckm = df['Dcrat(m)'] / 1000

plt.plot(Dckm, Dikm)
```

```python
df['log(Dc)'] = np.log(df['Dcrat(m)'])
df['log(Di)'] = np.log(df['#Dimp(m)'])
```

```python
xnew = df['log(Dc)'].values
ynew = df['log(Di)'].values
interp = interp1d(xnew, ynew, fill_value='extrapolate')
```

```python
np.exp(interp(np.log(200000))) #crater size as input; impactor size as output
```
